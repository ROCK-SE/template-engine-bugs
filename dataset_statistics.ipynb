```python
import pandas as pd

questions = pd.read_csv("data/so_questions.csv")
bugs = pd.read_excel("data/so_bugs.xlsx")
```

```python
# Table 1
question_count = {}
for row in questions["template-engine"]:
    for te in row.split(";"):
        question_count[te] = question_count.get(te, 0) + 1
summary = bugs[["Template Engine", "Language"]].value_counts().reset_index(name="# Bugs")
summary["# Questions"] = summary["Template Engine"].map(question_count)
summary[summary.columns[[0, 1, 3, 2]]]
```

```python
# Data Collection Process
post_2020 = questions[pd.to_datetime(questions["creation_time"]) > "2020-01-01"]
print(f"After 2020 cutoff filtering: {len(post_2020)} questions")
has_acc = post_2020[post_2020["has_acc_ans"] == 1]
print(f"After accepted answer existence filtering: {len(has_acc)} questions")
has_code = has_acc[has_acc["has_code"] == 1]
print(f"After code existence filtering: {len(has_code)} questions")
pos_score = has_code[has_code["score"] > 0]
print(f"After positive score filtering: {len(pos_score)} questions")
candidate_questions = pos_score[pos_score["has_bug_keywords"] == 1]
print(f"After bug keywords filtering: {len(candidate_questions)} questions")
```
